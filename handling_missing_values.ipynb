```python
import pandas as pd
```

```python
data = {
    'Name': ['manoj', 'madhu', 'manu', 'vaishu', 'abhi', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print("Mean Age:",df['Age'].mean())
```

```python
df['Age']= df['Age'].fillna(df['Age'].mean())
```

```python
print(df)
```

```python
data = {
    'Name': ['pushpa', 'pragathi', 'aishu', 'vaishu', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}

df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df['Salary']= df['Salary'].fillna(df['Salary'].median())
```

```python
print(df)
```

```python
data = {
    'Name': ['pushpa', 'pragathi', 'aishu', 'vaishu', 'Meena', 'Arjun'],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
}
df = pd.DataFrame(data)
print("Before Mode Impulation:")
print(df)
    
```

```python
print("Mode:",df['Department'].mode()[0])
```

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])

print("\nAfter Mode Imputation:")
print(df)
```

```python
data = {
    'Name': ['pushpa', 'pragathi', 'aishu', 'vaishu', 'Meena', 'Arjun'],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', None],
}
df = pd.DataFrame(data)
print("Before Forward Fill:")
print(df)
```

```python
df['Department'] = df['Department'].ffill()
print("\n After Forward Fill:")
print(df)
```
